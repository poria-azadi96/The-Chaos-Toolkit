```python
# Environment setup & Imports
import os, sys
sys.path.append(os.path.abspath('..'))
import numpy as np
import matplotlib.pyplot as plt
import ipywidgets as widgets
from IPython.display import display, HTML

from shared_utils.theme import PALETTE, apply_sleek_theme, create_dashboard_header
from pathological_functions import weierstrass_function, box_counting_dimension, generate_devils_staircase
from linear_phase_space import compute_spectral_properties, simulate_orbit_comparison

apply_sleek_theme()
display(HTML(create_dashboard_header('ACT 01: THE CRISIS IN NEWTON\'S PARADISE', 'Interactive GUI Suite: Linear State Space & Pathological Fractals')))
```

```python
# Interactive Weierstrass Zoom GUI
def render_weierstrass(zoom_level, a, b):
    scale = 1.0 / (10.0 ** (zoom_level - 1))
    x = np.linspace(-scale, scale, 3000)
    y = weierstrass_function(x, a=a, b=b, n_terms=60)
    dim = 2.0 + (np.log(a) / np.log(b))
    fig, ax = plt.subplots(figsize=(10, 4), dpi=100)
    ax.plot(x, y, color=PALETTE['cyan'], lw=1.2)
    ax.set_title('Weierstrass Zoom: 10^{}x | Fractal Dim: {:.3f}'.format(zoom_level-1, dim))
    ax.set_xlim(-scale, scale)
    plt.show()

widgets.interactive(render_weierstrass, zoom_level=(1, 6, 1), a=(0.1, 0.9, 0.05), b=(3, 25, 2))
```

```python
# Interactive Linear State-Space & Phase Plane GUI
def render_oscillator(damping):
    props = compute_spectral_properties(m=1.0, k=4.0, b=damping)
    t, num, ana = simulate_orbit_comparison(m=1.0, k=4.0, b=damping)
    fig, axs = plt.subplots(1, 2, figsize=(12, 4.5), dpi=100)
    axs[0].plot(t, num[0], color=PALETTE['cyan'], lw=1.5, label='Numerical')
    axs[0].plot(t, ana[0], color=PALETTE['amber'], linestyle='--', label='Matrix Exp')
    axs[0].set_title('Position x(t) | {}'.format(props['morphology']))
    axs[0].legend()
    axs[1].plot(num[0], num[1], color=PALETTE['cyan'], lw=1.5)
    axs[1].scatter([0], [0], color=PALETTE['crimson'], s=40)
    axs[1].set_title('Phase Portrait | Trace: {:.2f}, Det: {:.2f}'.format(props['trace'], props['det']))
    plt.tight_layout()
    plt.show()

widgets.interactive(render_oscillator, damping=(0.0, 5.0, 0.1))
```
